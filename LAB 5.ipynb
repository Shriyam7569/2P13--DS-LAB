```python
import pandas as pd

roll_number = input("Enter your college roll number: ")

last_two = roll_number[-2:]

digit1 = int(last_two[0])
digit2 = int(last_two[1])

categories = ["billing", "account", "general"]

def create_entry(digit):
    category = categories[digit % 3]

    if category == "billing":
        question = "How can I pay my bill?"
        keywords = "bill payment fee charges"
        answer = "You can pay your bill using UPI, card, or net banking."

    elif category == "account":
        question = "How do I update my registered mobile number?"
        keywords = "mobile number update account"
        answer = "Go to Settings > Profile > Update Mobile Number."

    else:
        question = "What are your customer service hours?"
        keywords = "working hours support help"
        answer = "Our customer service is available from 9 AM to 5 PM."

    return {
        "question": question,
        "answer": answer,
        "keywords": keywords,
        "category": category
    }

fixed_entries = [
    {
        "question": "What is the annual fee?",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "How to reset password?",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "What are your working hours?",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "How can I pay the fee?",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment UPI fee",
        "category": "billing"
    }
]

personalized_entry1 = create_entry(digit1)
personalized_entry2 = create_entry(digit2)

faq_entries = fixed_entries + [
    personalized_entry1,
    personalized_entry2
]

df = pd.DataFrame(faq_entries)

print("\nQ1: Final DataFrame")
print(df)


def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for index, row in df.iterrows():
        text = (
            str(row["question"]) + " " +
            str(row["keywords"])
        ).lower()

        text_words = set(text.split())
        score = len(query_words.intersection(text_words))

        results.append({
            "question": row["question"],
            "answer": row["answer"],
            "category": row["category"],
            "score": score
        })

    return sorted(
        results,
        key=lambda x: x["score"],
        reverse=True
    )


print("\nQ2: Search FAQ")

query = input("Enter your query: ")
results = score_query(query, df)

for result in results:
    if result["score"] > 0:
        print(
            f"Score: {result['score']} | "
            f"Category: {result['category']} | "
            f"Question: {result['question']}"
        )


def same_category(category_name, df):
    return df[df["category"].str.lower() == category_name.lower()]


print("\nQ3: Same Category")

category_name = personalized_entry1["category"]

print("Selected category:", category_name)
print(same_category(category_name, df)[["question", "category"]])


print("\nQ4: Add Keyword")

for i in range(len(df)):
    print(i, "-", df.loc[i, "question"])

entry_number = int(input("Enter entry number (0-5): "))
new_keyword = input("Enter a new keyword: ")

df.loc[entry_number, "keywords"] = (
    df.loc[entry_number, "keywords"] + " " + new_keyword
)

df.to_csv("your_roll_number_faq_data.csv", index=False)

print("\nUpdated entry:")
print(df.loc[entry_number])
print("\nFile saved as: your_roll_number_faq_data.csv")


print("\nQ5: FAQ Count Per Category")

category_counts = df.groupby("category").size()
print(category_counts)


def best_matches(query, df):
    results = score_query(query, df)
    highest_score = results[0]["score"]

    return [
        result for result in results
        if result["score"] == highest_score
    ]


print("\nQ6: Tie Handling")

test_query = "fee"

print("Test query:", test_query)

best_results = best_matches(test_query, df)

for result in best_results:
    print(
        f"Score: {result['score']} | "
        f"Category: {result['category']} | "
        f"Question: {result['question']} | "
        f"Answer: {result['answer']}"
    )
```
